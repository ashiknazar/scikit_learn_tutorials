## Ridge Regression Calculation

Ridge Regression minimizes the following cost function:

$ J(\beta) = ||Y - X\beta||^2 + \lambda ||\beta||^2 $

where:
- $ X $ is the feature matrix  
- $ Y $ is the target vector  
- $ \beta $ is the coefficient vector  
- $ \lambda $ is the regularization parameter  
- $ ||\beta||^2 = \sum \beta_j^2 $ is the L2 norm of the coefficients  

The closed-form solution for the Ridge Regression coefficients is:

$ \beta = (X^TX + \lambda I)^{-1} X^TY $

### Example Calculation

Given:

$ X =
\begin{bmatrix}
1 & 2 \\
2 & 3 \\
3 & 4
\end{bmatrix},
\quad
Y =
\begin{bmatrix}
3 \\
5 \\
7
\end{bmatrix} $

For $ \lambda = 1 $, we calculate:

$ \beta_1 = 0.8, \quad \beta_2 = 1.1 $

Thus, the Ridge Regression equation is:

$ \hat{Y} = 0.8X_1 + 1.1X_2 $

This helps prevent overfitting by shrinking the coefficients compared to ordinary least squares.
