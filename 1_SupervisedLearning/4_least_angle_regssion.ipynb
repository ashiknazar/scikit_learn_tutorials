- Least Angle Regression (LARS) is an efficient regression algorithm that works well when the number of predictors (features) is much larger than the number of observations (samples). 

- Starting with all coefficients as zero.
- Finding the predictor most correlated with the residual.
- Moving the coefficient in that direction until another predictor becomes equally correlated.
- Then adjusting both coefficients together until another predictor reaches their level of correlation.
- Repeating this process until stopping conditions are met.

# Least Angle Regression (LARS) - Step-by-Step Example

## 📌 Problem Setup
We have the following standardized dataset:

| Sample | $X_1$ | $X_2$ | $X_3$ | $X_4$ | $X_5$ | $y$ |
|--------|---------|---------|---------|---------|---------|---------|
| 1      |  0.5   |  -0.3   |  0.8    | -0.2    |  0.1    |  1.2    |
| 2      | -0.2   |  0.4    | -0.5    |  0.7    |  0.3    | -0.8    |
| 3      |  0.3   |  0.2    |  0.4    | -0.6    | -0.4    |  0.5    |
| 4      | -0.7   |  0.5    | -0.9    |  0.3    |  0.2    | -1.5    |
| 5      |  0.6   | -0.1    |  0.7    | -0.4    | -0.5    |  1.0    |

Since our dataset is already standardized (mean = 0, standard deviation = 1 for each feature), we proceed to the next step.

---

## 📌 Step 1: Compute Correlation Between Features and $y$

We compute the absolute correlation between each feature $X_j$ and the response $y$:

$$
C_j = \left| \frac{\sum_{i=1}^{n} (X_{ij} - \bar{X}_j)(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n} (X_{ij} - \bar{X}_j)^2} \sqrt{\sum_{i=1}^{n} (y_i - \bar{y})^2}} \right|
$$

where:
- $X_{ij}$ is the $i$-th observation of feature $X_j$.
- $\bar{X}_j$ is the mean of feature $X_j$.
- $y_i$ is the $i$-th target value.
- $\bar{y}$ is the mean of $y$.

| Feature | Correlation $|C_j|$ |
|---------|--------------|
| $X_1$ | **0.85** |
| $X_2$ | 0.42 |
| $X_3$ | 0.76 |
| $X_4$ | 0.30 |
| $X_5$ | 0.62 |

🔹 **Step 1 Decision:** The feature with the highest correlation is **$X_1$ (0.85)**, so we start with it.

---

## 📌 Step 2: Move Towards $X_1$
We increase the coefficient $\beta_1$ of $X_1$ in the direction that reduces the residual error.

$$
r = y - \hat{y} = y - \beta_1 X_1
$$

We move $\beta_1$ in the direction of $X_1$ **until another predictor reaches the same absolute correlation as $X_1$ with the residual**.

---

<div style="border: 2px solid black; padding: 10px; background-color: #f9f9f9; border-radius: 5px;">
<b>what’s happening step by step:</b>  

### Updating the coefficient $( \beta_1 )$:  
- We adjust $( \beta_1 )$ in the direction that reduces the residual $( r = y - \hat{y} )$.  
- This means that we increase $( \beta_1 )$ to better fit $( X_1 ) to ( y )$, reducing the prediction error.  

### Change in correlation with residuals:  
- As we update $( \beta_1 )$,the residual $( r )$ changes.  
- The correlation between $( X_1 )$ and the residual decreases because $( X_1 )$ is already being accounted for in the prediction.  

### Meeting another feature’s correlation:  
- We continue adjusting $( \beta_1 )$ until another predictor $( X_2 )$ (or any other feature) has the same absolute correlation with the new residual.  
- This is because the residual now contains the remaining variance that $( X_1 )$ could not explain, and another feature might explain it just as well as $( X_1 )$.  
- At this point, the algorithm will consider bringing $( X_2 )$ into the model.  

### Conclusion:  
This is the core idea behind **Least Angle Regression (LARS)**—we move along the direction of the most correlated predictor until another feature becomes equally correlated with the residual. Then, we adjust both together.  

So yes, reducing the residual by updating the coefficient naturally decreases the correlation of that feature with the residual, and the only way another feature can meet it is if its correlation stays constant (or decreases more slowly).  
</div>


---


## 📌 Step 3: Introduce the Next Feature
As $\beta_1$ increases, the correlations of other features with the residual change. Suppose $X_3$ now reaches the same correlation as $X_1$ (0.76).

🔹 **Step 3 Decision:** Now we include $X_3$ and move both $\beta_1$ and $\beta_3$ together in a **least-angle direction**.

---

## 📌 Step 4: Move Along the Least Angle Direction
Instead of increasing $\beta_1$ and $\beta_3$ separately, LARS moves in the **joint least-angle direction**.

### 🔹 Compute the Equiangular Vector
The equiangular vector $\mathbf{u}$ is:

$$
\mathbf{u} = X_A w
$$

where:
- $X_A$ is the matrix of selected predictors.
- $w$ is a weight vector computed as:

$$
w = (X_A^T X_A)^{-1} \mathbf{1}
$$

- $\mathbf{1}$ is a vector of ones.

### 🔹 Compute the Step Size
The step size $\gamma$ is:

$$
\gamma = \min \left\{ \frac{C - C_A}{a - a_A}, \frac{C + C_A}{a + a_A} \right\}
$$

where:
- $C$ is the current correlation.
- $C_A$ is the correlation of active variables.
- $a$ and $a_A$ are directional derivatives.

---

## 📌 Step 5: Continue Adding Features
As $\beta_1$ and $\beta_3$ increase, residual $r$ changes. When another feature (say $X_5$, correlation 0.62) reaches the same level of correlation, it gets added, and we move along the least-angle direction of **$X_1, X_3, X_5$**.

We repeat this process until:
1. **All significant features are added** OR  
2. **We reach a stopping condition** (such as Lasso's constraint).

---

## 📌 Step 6: Interpret the Final Model
After a few steps, we might obtain a model like:

$$
\hat{y} = 0.6 X_1 + 0.4 X_3 + 0.3 X_5
$$

This means:
- $X_1$, $X_3$, and $X_5$ are **the most important predictors**.
- $X_2$ and $X_4$ were **not included**, indicating they do not strongly contribute to $y$.

---

## 📌 Step 7: Compare with Lasso
LARS is closely related to **Lasso regression**, but instead of applying a penalty term, it follows a path that **only includes important features**.

Lasso solves:

$$
\min_{\beta} \sum_{i=1}^{n} \left( y_i - \sum_{j=1}^{p} X_{ij} \beta_j \right)^2 + \lambda \sum_{j=1}^{p} |\beta_j|
$$

while LARS **finds the same solution but without explicitly requiring $\lambda$ selection**.

---

## **💡 Key Takeaways**
✅ **LARS selects features step by step** like forward selection but in a more controlled manner.  
✅ **It adjusts coefficients gradually**, making it computationally efficient.  
✅ **It gives similar results to Lasso**, but without needing cross-validation for a penalty parameter.  
✅ **It finds important features while keeping coefficients small**, which helps prevent overfitting.
