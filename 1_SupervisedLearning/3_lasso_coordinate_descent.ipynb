- Lasso Regression differs from Ridge because it uses L1 regularization, which can shrink some coefficients to exactly zero, effectively performing feature selection.

- While Ridge Regression has a **closed-form solution**, Lasso Regression doesn't. Instead, Lasso uses an **iterative optimization technique**, with Coordinate Descent being one of the most efficient methods.

### Why Coordinate Descent?

- Instead of solving for all coefficients at once, it updates one coefficient at a time while keeping others fixed.
- This is computationally efficient, especially when there are many features.
- It exploits sparsity: Lasso often results in zero coefficients, making it ideal for feature selection.

## Closed-Form Solution vs. Iterative Optimization

### **1. Closed-Form Solution**
A closed-form solution provides a **direct mathematical formula** to compute the optimal values of parameters. It is used in models where a solution can be derived algebraically.

#### **Example: Ordinary Least Squares (OLS) Regression**
The optimal coefficients $ \beta $ in OLS are given by:

$$ \beta = (X^TX)^{-1} X^TY $$

#### **Advantages**
- Fast and computationally efficient for small datasets.
- No need for iterative updates.
- Exact solution without approximations.

#### **Disadvantages**
- Requires matrix inversion, which is computationally expensive for large datasets.
- Doesn't work well when $ X^TX $ is singular or ill-conditioned.

---

### **2. Iterative Optimization Techniques**
Iterative optimization methods **start with an initial guess** and gradually improve the solution by minimizing a loss function step by step.

#### **Example: Gradient Descent (Used in Logistic Regression, Deep Learning)**
Gradient Descent updates parameters iteratively:

$$ \beta^{(t+1)} = \beta^{(t)} - \alpha \nabla J(\beta^{(t)}) $$

where:
- $ \alpha $ is the learning rate.
- $ \nabla J(\beta) $ is the gradient of the cost function.

#### **Example: Coordinate Descent (Used in Lasso Regression)**
Instead of updating all coefficients simultaneously, Coordinate Descent updates **one coefficient at a time** using the **soft-thresholding function**:

$$ \beta_j = S\left(\frac{X_j^T (Y - X\beta + X_j\beta_j)}{||X_j||^2}, \lambda\right) $$

#### **Advantages**
- Works well for large datasets where matrix inversion is infeasible.
- Can handle non-differentiable terms (e.g., L1 regularization in Lasso).
- Often more stable for high-dimensional problems.

#### **Disadvantages**
- Requires careful tuning of hyperparameters (e.g., learning rate in Gradient Descent).
- Can be slow to converge, depending on the optimization algorithm.
- May get stuck in local minima (especially in non-convex problems).

---

### **Comparison Table**

| Feature               | Closed-Form Solution          | Iterative Optimization         |
|----------------------|----------------------------|-------------------------------|
| **Computation**       | Direct formula            | Step-by-step updates          |
| **Speed**            | Fast for small datasets   | Efficient for large datasets  |
| **Accuracy**         | Exact solution            | Approximate solution          |
| **Scalability**      | Poor for large data       | Works well for big data       |
| **Usage**            | OLS, Ridge Regression     | Lasso, Logistic Regression, Deep Learning |

---

### **Conclusion**
- **Use a closed-form solution** when the problem is small and the matrix is well-conditioned.
- **Use an iterative optimization method** for large-scale problems or when closed-form solutions are infeasible.

### 2.1 Lasso Regression - First Iteration Residual Computation

Lasso regression solves the optimization problem:

$$
\hat{\beta} = \arg\min_{\beta} \left\{ ||y - X\beta||^2 + \lambda ||\beta||_1 \right\}
$$

where \( \lambda ||\beta||_1 \) is the **L1 regularization term** that promotes sparsity.

#### Step 1: Initial Residual

We start with an initial guess:

$$
\beta = 0
$$

Thus, the initial predicted values are:

$$
\hat{y} = X \beta = 0
$$

The residual at this stage is:

$$
r = y - \hat{y} = y
$$

#### Step 2: First Iteration Update (Coordinate Descent)

Lasso updates each coefficient $( \beta_j )$ one at a time. For the first feature $( X_1 )$:

$$
r_1 = y + X_1 \beta_1 - \hat{y}
$$

Since $( \hat{y} = 0 )$, this simplifies to:

$$
r_1 = y
$$

Thus, at the first iteration, the residual is simply equal to $( y ).$

### Lasso Regression - Coordinate Descent Step

#### Step 1: Compute Correlation

For each feature \( X_1 \), compute the correlation with the residual:

$$
\rho_1 = X_1^T r_1
$$

#### Step 2: Apply Soft-Thresholding

Update the coefficient \( \beta_1 \) using soft-thresholding:

$$
\beta_1 =
\frac{\rho_1 - \lambda/2}{X_1^T X_1}, \quad \text{if } \rho_1 > \lambda/2
$$

Else, shrink \( \beta_1 \) to zero.

#### Step 3: Repeat for All Features

Apply the same process for all coefficients:

$$
\beta_2, \beta_3, \beta_4, \dots
$$

This ensures Lasso's **L1 regularization** promotes sparsity by setting small coefficients to zero.

## **Step-by-Step Breakdown of Lasso Coordinate Descent**

### **1. Initialize Coefficients**
- Start with an initial guess, usually:
  $ \beta = 0 $
- Initial predicted values:
  $ \hat{y} = X\beta = 0 $
- Initial residual:
  $ r = y - \hat{y} = y $

🔹 **Why?**  
- Provides a neutral starting point for iterative updates.

---

### **2. Compute Correlation Between Each Feature and Residual**
- For each feature $ X_j $, compute its correlation with the residual:

  $$
  \rho_j = X_j^T r
  $$

  or, using mean-centered residuals:

  $$
  \rho_j = X_j^T (r - \bar{r})
  $$

  where:
  $ \bar{r} = \frac{1}{n} \sum_{i=1}^{n} r_i $

🔹 **Why?**  
- Determines how much each $ \beta_j $ should be updated based on its importance.

---

### **3. Apply Soft-Thresholding to Update $ \beta_j $**
- Instead of using ordinary regression update:

  $$
  \beta_j = \frac{\rho_j}{X_j^T X_j}
  $$

- Lasso applies **soft-thresholding**:

  $$
  \beta_j =
  \frac{\rho_j - \lambda/2}{X_j^T X_j}, \quad \text{if } \rho_j > \lambda/2
  $$

  Else, shrink $ \beta_j $ to zero if $ |\rho_j| \leq \lambda/2 $:

  $$
  \beta_j = 0
  $$

🔹 **Why?**  
- Enforces **sparsity** by eliminating small coefficients.

---

### **4. Update Residuals**
- After updating $ \beta_j $, compute new residual:

  $$
  r = y - X \beta
  $$

🔹 **Why?**  
- Ensures subsequent updates use the latest predictions.

---

### **5. Repeat for All Features Until Convergence**
- Iterate over all features until coefficient updates are negligible.

🔹 **Why?**  
- Finds the best sparse solution through multiple passes.

---

| Step | Purpose |
|------|---------|
| **Initialize $ \beta $** | Start with a neutral state |
| **Compute correlation $ \rho_j $** | Identify which features relate most to residuals |
| **Apply soft-thresholding** | Shrink small coefficients to zero for sparsity |
| **Update residuals** | Ensure next updates reflect the latest predictions |
| **Repeat until convergence** | Guarantee optimal coefficient selection |

# Lasso Coordinate Descent - Step-by-Step

## **1. Initialize Coefficients**
- Set all coefficients to zero: $ \beta = 0 $.
- Compute initial residual: $ r = y $.

## **2. Iterate Over Each Feature $ X_j $ (One at a Time)**
- Fix all other coefficients and update only $ \beta_j $.

## **3. Compute Partial Residual**
- Remove the contribution of other features:  
  $$ r_j = y - \sum_{k \neq j} X_k \beta_k $$

## **4. Compute Correlation Between Feature and Residual**
- Compute:  
  $$ \rho_j = X_j^T r_j $$
- This measures how much the feature $ X_j $ is related to the residual error.

## **5. Apply Soft-Thresholding to Update $ \beta_j $**
- If $ \rho_j > \lambda/2 $:
  $$ \beta_j = \frac{\rho_j - \lambda/2}{X_j^T X_j} $$
- If $ \rho_j < -\lambda/2 $:
  $$ \beta_j = \frac{\rho_j + \lambda/2}{X_j^T X_j} $$
- Otherwise, set $ \beta_j = 0 $.

## **6. Update Residuals**
- After updating $ \beta_j $, update the residual:  
  $$ r = y - X \beta $$

## **7. Repeat Steps 2-6 Until Convergence**
- Iterate over all features multiple times until coefficient updates are small.
