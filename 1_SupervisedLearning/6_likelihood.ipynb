## Likelihood in Machine Learning
- Likelihood is a fundamental concept in machine learning and statistics. It measures how probable a set of observations (data) is, given a particular model and its parameters.


## 1. Likelihood vs. Probability
- **Probability**: Given a known model, what is the probability of seeing a particular outcome?
  - Example: If we flip a fair coin, the probability of getting heads is 0.5.


- **Likelihood**: Given observed data, what is the most likely model (or parameters) that could have generated it? 
   - Example: If we flip a coin 10 times and get 8 heads, what is the likelihood that the coin is biased (e.g., p=0.8)?

- Probability → Predicts data from parameters
- Likelihood → Infers parameters from data

## 2. Likelihood Function
- The likelihood function is used in statistical inference to estimate model parameters.


## 4. Maximum Likelihood Estimation (MLE)
MLE is a method to find the parameters θ that maximize the likelihood function.
- Define the likelihood function L(θ).
- Take the logarithm (log-likelihood) to simplify calculations.
- Differentiate log-likelihood with respect to parameters.
- Set derivatives to zero and solve for parameters.
