- AIC and BIC are model selection criteria used to compare different statistical models and choose the one that best balances goodness of fit and complexity.

### 1. Akaike Information Criterion (AIC)
AIC measures how well a model fits the data while penalizing complexity to prevent overfitting. The formula is:
- AIC=2k−2ln(L)
- k = number of parameters in the model

- L = maximum likelihood of the model

- Lower AIC values indicate a better model.


### 2. Bayesian Information Criterion (BIC)
BIC is similar to AIC but applies a stronger penalty for model complexity, making it stricter in favoring simpler models. The formula is:

- BIC=kln(n)−2ln(L)

- n = number of data points
- k = number of parameters in the model
- L = maximum likelihood of the model

- Lower BIC values indicate a better model.


## Likelihood

- Likelihood is a fundamental concept in statistics and machine learning that measures how well a given model explains observed data. It tells us how probable it is to observe the given data under a specific model with certain parameter values.

<div style="border: 2px solid black; padding: 10px; background-color: #f9f9f9; border-radius: 5px;">

### **What is Likelihood?**  

**Likelihood** measures how well a given model explains observed data. It represents the probability of the data given specific model parameters.  

Mathematically, for a dataset $ \mathcal{D} = \{ x_1, x_2, \dots, x_n \} $ and a model parameterized by $ \theta $, the **likelihood function** is:  

$$ L(\theta) = P(\mathcal{D} | \theta) $$  

---

### **Likelihood vs. Probability**  
- **Probability:** Given a model and parameters, what is the chance of a new event occurring?  
- **Likelihood:** Given observed data, what are the most likely parameters for the model?  

---

### **Example: Likelihood in a Coin Toss**  
Suppose we flip a biased coin $ n = 10 $ times and observe $ k = 7 $ heads. If the true probability of heads is $ p $, then the likelihood function is:  

$$ L(p) = P(k=7 | p) = \binom{10}{7} p^7 (1-p)^3 $$  

This function tells us how likely different values of $ p $ are, given that we observed 7 heads.  

---

### **Maximum Likelihood Estimation (MLE)**  
MLE finds the parameter $ \theta $ that **maximizes the likelihood**:  

$$ \hat{\theta} = \arg\max_{\theta} L(\theta) $$  

For example:  
- In a normal distribution $ X \sim \mathcal{N}(\mu, \sigma^2) $, MLE estimates $ \mu $ as the sample mean and $ \sigma^2 $ as the sample variance.  
- In logistic regression, MLE finds the best coefficients that maximize the likelihood of the observed labels given the features.  

---

### **Log-Likelihood**  
Since likelihood functions often involve products of probabilities, we use the **log-likelihood function** to simplify computations:  

$$ \log L(\theta) = \sum_{i=1}^{n} \log P(x_i | \theta) $$  

- Log-likelihood avoids numerical issues by converting products into sums.  
- It is used in logistic regression, neural networks, and Bayesian inference.
