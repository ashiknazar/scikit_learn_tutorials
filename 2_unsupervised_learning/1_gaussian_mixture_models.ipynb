## Gaussian mixture models



A **Gaussian Mixture Model (GMM)** is a probabilistic model that assumes data is generated from a mixture of multiple Gaussian distributions.

## **Key Features**
- **Soft Clustering**:  Unlike k-means, which assigns each data point to a single cluster, GMM provides probabilities for a point belonging to different clusters.Assigns probabilities to cluster memberships.
- **Flexible Shapes**: Models elliptical clusters, unlike k-means.
- **Elliptical Clusters**: GMM can model elliptical clusters (not just spherical ones like k-means).
- **Density Estimation**:It estimates the probability density function of the data, making it Useful for anomaly detection and generative modeling.

## how it works
- Each cluster is modeled as a Gaussian (Normal) distribution with its own mean and covariance.
- The Expectation-Maximization (EM) algorithm is used to:
    - **Expectation Step (E-step)**: Compute the probability that each data point belongs to each cluster
    - **Maximization Step (M-step)**: Update the parameters (mean, covariance, and weights) of the Gaussians to maximize the likelihood.



## **Implementation in Python**
```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Generate sample data
X = np.concatenate([
    np.random.normal([0, 0], 0.5, (100, 2)),
    np.random.normal([3, 3], 0.5, (100, 2))
])

# Fit GMM with 2 clusters
gmm = GaussianMixture(n_components=2, covariance_type='full', random_state=42)
gmm.fit(X)

# Predict cluster assignments
labels = gmm.predict(X)

# Plot results
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=30)
plt.scatter(gmm.means_[:, 0], gmm.means_[:, 1], c='red', marker='x', label='Centroids')
plt.legend()
plt.show()


- When to Use GMM?
   - When clusters have different shapes and orientations.
   - When you need probabilistic clustering (soft clustering).
   - For anomaly detection in datasets.

___
